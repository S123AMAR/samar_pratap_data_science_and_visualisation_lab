```python
arr = np.array([-16,25,-36,49])
absolute = np.abs(arr)
result = np.sqrt(absolute)
print("Absolute value:", absolute)
print("Square roots:", result)
```

pandas series

```python

books = pd.Series(
    [24, 31,19, 27],
    index=['Monday','Tuesday', 'Wednesday','Thursday'])
print(books)
```

```python

prices = pd.Series(
    [45000, 18000, 22000],
    index=['Laptop', 'Tablet', 'Phone']
)

print(prices['Tablet'])
```

```python

rainfall = pd.Series([12, 0, 18, 5,23])
print(rainfall[rainfall >10])
```

```python

data= {'Gold': 7200, 'Silver': 95, 'Copper': 850}
s = pd.Series(data)
print("Index:")
print(s.index)

print("Values:")
print(s.values)
```

```python

ratings= pd.Series([3.5, 4.2, 2.8, 4.7])
ratings.name ='Customer Ratings'
ratings.index.name = 'Review ID'
print(ratings)
```

```python

 s= pd.Series([11, 22,33], index=['X1', 'X2','X3'])
print("X2 exists:",'X2' in s.index)
print("X4 is absent:",'X4' not in s.index)
```

```python

stock =pd.Series(
    [16, 28,12],
    index=['Keyboard', 'Mouse', 'Monitor'])
result =stock.to_dict()

print(result)
```

```python

data= {'Hindi':78,'English':84, 'Computer': 91}
s =pd.Series(data, index=['Computer', 'Hindi', 'Art', 'English'])

print(s)
```

```python

expenses=pd.Series([1200,None, 850,None, 600])
print("isnull():")
print(expenses.isnull())
print("\nnotnull():")
print(expenses.notnull())
```
