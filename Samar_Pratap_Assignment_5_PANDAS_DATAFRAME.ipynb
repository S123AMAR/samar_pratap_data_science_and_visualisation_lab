```python
import pandas as pd

data={
    'Name' : ['Mohan', 'Riya', 'Rahul', 'Priya'],
    'Age' : [20, 21, 19, 18],
    'City' : ['Mathura', 'Agra', 'Noida', 'Delhi']
}
student_df = pd.DataFrame(data)
print(student_df)
```

```python
data = {
    'Book': ['Python','Java', 'C++'],
    'Author': ['Amit','Rahul','Neha'],
    'Price': [400,500, 350]
}
book_df = pd.DataFrame(data)
print(book_df)
```

```python
movie_df = pd.DataFrame({
    'Movie':['3Idiots','Dangal', 'Pathaan'],
    'Year': [2009, 2016,2023],
    'Rating': [8.4, 8.3,5.8]})
print(movie_df)
```

```python
new_df= movie_df[['Movie','Rating']].copy()
new_df['Director'] =['Rajkumar Hirani', 'Nitesh Tiwari','Siddharth Anand']
print(new_df)
```

```python
cities= pd.Series({
    0: 'Delhi',
    3: 'Mumbai'})
print(cities)
```

```python
df= pd.DataFrame(index=range(4))
df['Branch'] = cities
print(df)
```
