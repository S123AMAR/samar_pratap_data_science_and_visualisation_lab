```python
import pandas as pd
s = pd.Series([90, 75,82],index=['Ravi','Aman', 'Neha'])
d = s.to_dict()
s2 = pd.Series(d)
print("Original Series:")
print(s)
print("\nDictionary:")
print(d)

print("\nNew Series:")
print(s2)
```

```python
import pandas as pd

sales = pd.Series({
    'Laptop': 800,
    'Mobile': 450,
    'Tablet': 650,
    'Keyboard': 300
})
print("Sales Values:")
print(sales.values)

print("\nIndex:")
print(sales.index)
print("\nLaptop Sales:")
print(sales['Laptop'])
print("\nSales Above 500:")
print(sales[sales > 500])
print("\nIs Mobile Present?")
print('Mobile' in sales.index)


print("\nDictionary:")
print(sales.to_dict())
```

```python
import pandas as pd

data = {'A': 15,'B': 35,'C': 25}
s = pd.Series(data, index=['C', 'A','D', 'B'])
s.name = 'Observations'
s.index.name = 'Category'
print(s)
print("\nBoolean Filter:")
print(s[s > 20])
print("\nNot Null:")
print(pd.notnull(s))
```
